````python
df["Year"] = df["Order Date"].dt.year

df["Month"] = df["Order Date"].dt.month_name()
````

````python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

df["Month"] = pd.Categorical(df["Month"],
                             categories=month_order,
                             ordered=True)
````

````python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales
````

````python
plt.figure(figsize=(12,6))

plt.plot(monthly_sales.index,
         monthly_sales.values,
         marker="o",
         linewidth=3)

plt.title("Monthly Sales Trend", fontsize=16)

plt.xlabel("Month")

plt.ylabel("Total Sales ($)")

plt.grid(True, linestyle="--", alpha=0.5)

plt.xticks(rotation=45)

plt.tight_layout()

plt.show()
````

# Correlation Heatmap

## Objective

Visualize the correlation between the numerical variables in the Superstore dataset to understand how Sales, Profit, Quantity, and Discount are related.

## Visualization Used

- **Chart Type:** Correlation Heatmap
- **Variables:** Sales, Profit, Quantity, Discount

## Observation

The heatmap displays the correlation coefficients between the numerical variables.

| Variables | Correlation | Interpretation |
|-----------|------------:|----------------|
| Sales ↔ Profit | **0.48** | Moderate Positive Correlation |
| Sales ↔ Quantity | **0.20** | Weak Positive Correlation |
| Sales ↔ Discount | **-0.03** | Very Weak Negative Correlation |
| Profit ↔ Quantity | **0.07** | Very Weak Positive Correlation |
| Profit ↔ Discount | **-0.22** | Weak Negative Correlation |
| Quantity ↔ Discount | **0.01** | Almost No Correlation |

## Business Insight

- **Sales and Profit** have a **moderate positive correlation (0.48)**, indicating that higher sales generally lead to higher profits, although other factors also influence profitability.
- **Discount and Profit** have a **negative correlation (-0.22)**, suggesting that larger discounts tend to reduce overall profit.
- **Sales and Quantity** show a weak positive relationship (0.20), meaning selling more units contributes to higher sales but is not the only influencing factor.
- **Quantity and Discount** have almost no relationship, indicating discounts are applied independently of the number of units sold.

## Conclusion
